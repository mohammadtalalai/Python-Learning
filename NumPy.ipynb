# 21. Practical Example — Student Scores

Let's use NumPy to perform a simple statistical analysis of student scores.

````python
scores = np.array([
    85, 90, 78, 92, 88,
    76, 95, 89, 84, 91,
    73, 87, 94, 82, 79
])
````

**Basic Information**

````python
print("Number of Students:", scores.size)
print("Minimum Score:", np.min(scores))
print("Maximum Score:", np.max(scores))
print("Average Score:", np.mean(scores))
print("Median Score:", np.median(scores))
````

**Variability**

````python
print("Standard Deviation:", np.std(scores))
print("Variance:", np.var(scores))
````

**Quartiles**

````python
q1 = np.percentile(scores, 25)
q2 = np.percentile(scores, 50)
q3 = np.percentile(scores, 75)

print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
````

## Outlier Detection Using IQR

The Interquartile Range:

IQR = Q3 - Q1

Lower Bound:

Q1 - 1.5 × IQR

Upper Bound:

Q3 + 1.5 × IQR

````python
q1 = np.percentile(scores, 25)
q3 = np.percentile(scores, 75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print("IQR:", iqr)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
````

**Find Outliers**

````python
outliers = scores[
    (scores < lower_bound) |
    (scores > upper_bound)
]

outliers
````
